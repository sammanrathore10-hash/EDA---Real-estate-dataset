```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_csv('/content/House Price India.csv')
df
```

```python
# Q1. What is the average, median, and standard deviation of house prices?
print(df['Price'].mean())
print(df['Price'].median())
print(df['Price'].std())
```

```python
# Q2. Which number of bedrooms is most common?
df.groupby('number of bedrooms')['number of bedrooms'].count().sort_values(ascending=False)
```

```python
# Q3. Check if the price data is skewed.
df['Price'].plot(kind='hist')
print('Price Skewness :', df['Price'].skew())
```

```python
# Q4. What is the average price for each number of bedrooms?
df.groupby('number of bedrooms')['Price'].mean().sort_values(ascending=False)
```

```python
# Q5.  What is the relationship between living area and price? (Use : Area without Basement).
df.plot(x='Price', y='Area of the house(excluding basement)', kind='scatter')
df[['Price','Area of the house(excluding basement)']].corr()
```

```python
# Q6. Identify any anomalies where houses have high prices but low areas
df.plot(x='Price', y='living area', kind='scatter')
df[['Price','living area']].corr()
```

```python
# Q7. Compare average price based on number of floors and houses with or without waterfront.
df.groupby(['number of floors','waterfront present'])['Price'].mean()
```

```python
# Q8. Identify the minimum and maximum house price. What does this indicate?
print('Minimum House Price :', df['Price'].min())
print('Maximum House Price :', df['Price'].max())
```

```python
# Q9. Which location (zipcode/area) has the highest average price?
df.groupby('Postal Code')['Price'].mean().sort_values(ascending=False)
```
